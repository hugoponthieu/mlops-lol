```python
DATA_DIR = "../data"
```

```python
from dvc.api import DVCFileSystem  # type: ignore
from dotenv import load_dotenv

import pandas as pd
import os

load_dotenv()

repo = DVCFileSystem(
    os.environ["DVC_GIT_REPO"],
    remote="s3-storage",
    remote_config={
        "access_key_id": os.environ["DVC_S3_ACCESS_KEY"],
        "secret_access_key": os.environ["DVC_S3_SECRET_KEY"],
    },
)

with repo.open("data/raw/dataset.csv") as f:
    df = pd.read_csv(f)

df.to_csv(f"{DATA_DIR}/raw/dataset.csv")
```
