# 03 Preprocessing

Load engineered match features from artifacts, keep labeled matches, create chronological train/valid/test splits, and export modeling artifacts.

```python
from pathlib import Path
import json

import pandas as pd


def resolve_artifacts_dir() -> Path:
    cwd = Path.cwd().resolve()
    candidates = [cwd / "artifacts", cwd.parent / "artifacts"]
    for candidate in candidates:
        if candidate.exists():
            return candidate
    return cwd.parent / "artifacts" if cwd.name == "notebooks" else cwd / "artifacts"


ARTIFACTS_DIR = resolve_artifacts_dir()
ARTIFACTS_DIR.mkdir(parents=True, exist_ok=True)
```

```python
FEATURE_COLUMNS = [
    "elo_diff",
    "winrate_last_5_diff",
    "winrate_last_10_diff",
    "winrate_last_20_diff",
    "matches_played_diff",
    "days_since_last_match_diff",
    "head_to_head_winrate_diff",
    "blue_side_team_winrate",
    "red_side_team_winrate",
]

match_features = pd.read_csv(ARTIFACTS_DIR / "match_features.csv", parse_dates=["date"])
labeled_matches = match_features[match_features["blue_team_win"].notna()].copy()
labeled_matches["blue_team_win"] = labeled_matches["blue_team_win"].astype("Int64")
labeled_matches.head()
```

```python
def chronological_split_local(
    df: pd.DataFrame,
    train_frac: float = 0.7,
    valid_frac: float = 0.15,
):
    if not 0 < train_frac < 1:
        raise ValueError("train_frac must be between 0 and 1")
    if not 0 < valid_frac < 1:
        raise ValueError("valid_frac must be between 0 and 1")
    if train_frac + valid_frac >= 1:
        raise ValueError("train_frac + valid_frac must be less than 1")

    ordered = df.sort_values(["date"], kind="mergesort").reset_index(drop=True)
    n_rows = len(ordered)
    train_end = int(n_rows * train_frac)
    valid_end = int(n_rows * (train_frac + valid_frac))

    train_df = ordered.iloc[:train_end].reset_index(drop=True)
    valid_df = ordered.iloc[train_end:valid_end].reset_index(drop=True)
    test_df = ordered.iloc[valid_end:].reset_index(drop=True)
    return train_df, valid_df, test_df


train_df, valid_df, test_df = chronological_split_local(labeled_matches)
pd.DataFrame(
    {
        "split": ["train", "valid", "test"],
        "rows": [len(train_df), len(valid_df), len(test_df)],
        "start_date": [train_df["date"].min(), valid_df["date"].min(), test_df["date"].min()],
        "end_date": [train_df["date"].max(), valid_df["date"].max(), test_df["date"].max()],
    }
)
```

```python
split_frames = {"train": train_df, "valid": valid_df, "test": test_df}

for split_name, split_df in split_frames.items():
    split_df[FEATURE_COLUMNS].to_csv(ARTIFACTS_DIR / f"{split_name}_features.csv", index=False)
    split_df[["blue_team_win"]].to_csv(ARTIFACTS_DIR / f"{split_name}_labels.csv", index=False)

(ARTIFACTS_DIR / "feature_columns.json").write_text(json.dumps(FEATURE_COLUMNS, indent=2))

{
    "train_features": str(ARTIFACTS_DIR / "train_features.csv"),
    "valid_features": str(ARTIFACTS_DIR / "valid_features.csv"),
    "test_features": str(ARTIFACTS_DIR / "test_features.csv"),
    "train_labels": str(ARTIFACTS_DIR / "train_labels.csv"),
    "valid_labels": str(ARTIFACTS_DIR / "valid_labels.csv"),
    "test_labels": str(ARTIFACTS_DIR / "test_labels.csv"),
    "feature_columns": str(ARTIFACTS_DIR / "feature_columns.json"),
}
```
