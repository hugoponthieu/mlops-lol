# 02 Feature Engineering

Build chronological, leakage-safe team features from prior matches only.

```python
from pathlib import Path
import sys

ROOT = Path("..").resolve()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from mlops.data import load_matches
from mlops.features import build_match_features

DATA_PATH = ROOT / "data" / "matchs_stats.csv"
df = load_matches(DATA_PATH)
feature_df = build_match_features(df)
feature_df.head()
```

```python
feature_columns = [
    "elo_diff",
    "winrate_last_5_diff",
    "winrate_last_10_diff",
    "winrate_last_20_diff",
    "matches_played_diff",
    "days_since_last_match_diff",
    "head_to_head_winrate_diff",
    "blue_side_team_winrate",
    "red_side_team_winrate",
]
feature_df[feature_columns].describe().T
```

```python
feature_df[["date", "blue_team", "red_team", "blue_team_win", "elo_diff"]].head(10)
```
