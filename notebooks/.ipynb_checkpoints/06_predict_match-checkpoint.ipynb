# 06 Predict A Match

Train the baseline model on historical labeled matches, then predict the result of one future blue-vs-red matchup.

```python
from pathlib import Path
import sys

import pandas as pd

ROOT = Path("..").resolve()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from mlops.data import load_matches
from mlops.predict import predict_match_result

DATA_PATH = ROOT / "data" / "matchs_stats.csv"
history_df = load_matches(DATA_PATH)
history_df.head()
```

Edit the values below to predict any matchup. The model always returns the probability that the blue side wins, and the red-side probability is `1 - blue_win_prob`.

```python
match_date = "2022-01-01"
blue_team = "FunPlus_Phoenix"
red_team = "FunPlus_Phoenix"
event = "Main"
patch = "12.1"
season = 12
```

```python
prediction = predict_match_result(
    history_df=history_df,
    match_date=match_date,
    blue_team=blue_team,
    red_team=red_team,
    event=event,
    patch=patch,
    season=season,
)
pd.DataFrame([prediction])
```
